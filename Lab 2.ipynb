```python
%sql
SELECT date_format(order_purchase_timestamp, 'yyyy-MM') as order_month,
    order_status,
    COUNT(*) as total_orders
FROM orders_2025
GROUP BY order_month, order_status
ORDER BY order_month, order_status;
```
